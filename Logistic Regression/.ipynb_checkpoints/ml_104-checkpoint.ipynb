## Training, Validation and Test Sets

While building real-world machine learning models, it is quite common to split the dataset into three parts:

1. **Training set** - used to train the model, i.e., compute the loss and adjust the model's weights using an optimization technique. 


2. **Validation set** - used to evaluate the model during training, tune model hyperparameters (optimization technique, regularization etc.), and pick the best version of the model. Picking a good validation set is essential for training models that generalize well. [Learn more here.](https://www.fast.ai/2017/11/13/validation-sets/)


3. **Test set** - used to compare different models or approaches and report the model's final accuracy. For many datasets, test sets are provided separately. The test set should reflect the kind of data the model will encounter in the real-world, as closely as feasible.


<img src="https://i.imgur.com/j8eITrK.png" width="480">


As a general rule of thumb you can use around 60% of the data for the training set, 20% for the validation set and 20% for the test set. If a separate test set is already provided, you can use a 75%-25% training-validation split.


When rows in the dataset have no inherent order, it's common practice to pick random subsets of rows for creating test and validation sets. This can be done using the `train_test_split` utility from `scikit-learn`. Learn more about it here: https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html

To split the data scikit learn already have function for it

```python
from sklearn.model_selection import train_test_split
```

```python
train_val_df, test_df = train_test_split(weather_df,test_size=0.2,random_state=42)
# so in this we split the dataset into 80 , 20-> goes to Test set
```

```python
# now we split the remaining 80% of dataset into Training , validation set
train_df, val_df = train_test_split(weather_df,test_size=0.25,random_state=42)
```

```python
print("Train df ",train_df.shape)  
print("validation df ",val_df.shape)  
print("Test df ",test_df.shape) 

# the thing here is that since the rows in the splitted data set are random
# we could have a mixed dated datesets for training like having a row containing 2015 dataset
# in vaildation we have a row containing dataset for 2013 its like predicting past with future data
# which we dont want we need to predict future data with past 
# so this method wont work
# what we can do is split data in acc to year 
```

```python
plt.title('No. of Rows per Year')
sns.countplot(x=pd.to_datetime(weather_df.Date).dt.year);
```

```python
year = pd.to_datetime(weather_df.Date).dt.year
train_df = weather_df[year < 2015]
val_df = weather_df[year == 2015]
test_df = weather_df[year>2015]
```

```python
# now we will remove the RainTomorrow column from the training dataset
input_cols = list(train_df.columns)[1:-1]
target_col = ["RainTomorrow"]
```

```python
train_inputs = train_df[input_cols].copy() # create copy of train_df - rainTomorrow
target_input = train_df[target_col].copy() # create copy of RainTomorrow df
train_inputs
```

```python
val_inputs = val_df[input_cols].copy()
val_targets = val_df[target_col].copy()
```

```python
test_inputs = test_df[input_cols].copy()
test_targets = test_df[target_col].copy()
```

# now we will seperate catogorical columns from numeric column

```python
numeric_cols = train_inputs.select_dtypes(include=np.number).columns.tolist()
categorical_cols = train_inputs.select_dtypes("object").columns.tolist()
```

### Machine learning models cant work with missing numercial data, The process of filling the missing values is called imputation

there are many several techiniques for imputation , here we wil use the basic of called the simpleImputer that replace the missing values with the mean of the column

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(missing_values=np.nan,strategy="mean")
```

```python
weather_df[numeric_cols].isna().sum() # this function will return count of each columns for NaN
```

```python
imputer.fit(weather_df[numeric_cols])
```

```python
list(imputer.statistics_) # this returns a list of mean for each column
```

```python
train_inputs[numeric_cols] = imputer.transform(train_inputs[numeric_cols])
```

```python
val_inputs[numeric_cols] = imputer.transform(val_inputs[numeric_cols])
```

```python
test_inputs[numeric_cols] = imputer.transform(test_inputs[numeric_cols])
```

```python
train_inputs[numeric_cols].isna().sum()  # now there are no missing values
```

## Scaling Numeric Features

Another good practice is to scale numeric features to a small range of values e.g. $(0,1)$ or $(-1,1)$. Scaling numeric features ensures that no particular feature has a disproportionate impact on the model's loss. Optimization algorithms also work better in practice with smaller numbers.

The numeric columns in our dataset have varying ranges.

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(weather_df[numeric_cols])
```

```python
weather_df.describe()
```

```python
train_inputs[numeric_cols] = scaler.transform(train_inputs[numeric_cols])
val_inputs[numeric_cols] = scaler.transform(val_inputs[numeric_cols])
test_inputs[numeric_cols] = scaler.transform(test_inputs[numeric_cols])
```

## Encoding Categorical Data

Since machine learning models can only be trained with numeric data, we need to convert categorical data to numbers. A common technique is to use one-hot encoding for categorical columns.

<img src="https://i.imgur.com/n8GuiOO.png" width="640">

One hot encoding involves adding a new binary (0/1) column for each unique category of a categorical column. 

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
```

```python
encoder.fit(weather_df[categorical_cols])
```

```python
encoder.categories_
```

```python
encoded_cols = list(encoder.get_feature_names_out(categorical_cols))
encoded_cols
```

```python
train_inputs[encoded_cols] = encoder.transform(train_inputs[categorical_cols])
pd.set_option('display.max_columns', None)
```

```python
val_inputs[encoded_cols] = encoder.transform(val_inputs[categorical_cols])
test_inputs[encoded_cols] = encoder.transform(test_inputs[categorical_cols])
```

# saving the fine tunes dataset to the disk

```python
train_inputs.to_parquet("train_inputs.parquet",engine="fastparquet")
val_inputs.to_parquet("val_inputs.parquet",engine="fastparquet")
test_inputs.to_parquet("test_inputs.parquet",engine="fastparquet")
```
