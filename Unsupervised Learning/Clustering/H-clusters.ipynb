# Hierarical Clustering with iris data set

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
iris = datasets.load_iris()
```

```python
iris_data = pd.DataFrame(iris.data)
```

```python
iris_data.columns = iris.feature_names
```

```python
iris_data
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_scaled = scaler.fit_transform(iris_data)
x_scaled
```

```python
# pca is used for dimenstional reduction cause use dataset have 4 feature 
# we will be only be needing 2 features for clustering

from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components = 2)
```

```python
pca_scaled = pca.fit_transform(x_scaled)
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1])
```

```python
# now will apply clustering
# in this will be using Agglomerative Clustering
import scipy.cluster.hierarchy as sc
```

```python
plt.figure(figsize=(20,7))
plt.title("DengoGram")

sc.dendrogram(sc.linkage(pca_scaled,method="ward"))
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters = 2,metric="euclidean",linkage="ward")
cluster.fit(pca_scaled)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1],c=cluster.labels_)
```
