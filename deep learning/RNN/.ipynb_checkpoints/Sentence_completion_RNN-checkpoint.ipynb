# Sentence completion project

```python
import numpy as np
import pandas as pd
import  seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("qoute_dataset.csv")
```

```python
quotes = df["quote"]
```

```python
quotes = quotes.str.lower()
```

```python
import string
translator = str.maketrans("","",string.punctuation)
```

```python
quotes = quotes.apply(lambda x: x.translate(translator))
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocal_size = 10000
tokenizer = Tokenizer(num_words=vocal_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index = tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]  # this tell about the what word used the most 1 meaning the most used word
```

```python
sequence = tokenizer.texts_to_sequences(quotes)
```

```python
X = []
y = []
for seq in sequence:
    for i in range(1,len(seq)):
        input_seq = seq[:i]
        output_seq = seq[i]
        X.append(input_seq)
        y.append(output_seq)
```

```python

max_len = max(len(x) for x in X)
print(max_len)
```

```python

from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded = pad_sequences(X, maxlen=max_len, padding='pre')
```

```python

y = np.array(y)
```

```python

from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y, num_classes=vocal_size)
```

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,LSTM, Dense
```

```python

embedding_dim = 50
rnn_units = 128
```

```python
rnn_model = Sequential()

rnn_model.add(
    Embedding(input_dim=vocal_size, output_dim=embedding_dim, input_length=max_len)
)
rnn_model.add(SimpleRNN(units=rnn_units))
rnn_model.add(Dense(units=vocal_size, activation='softmax'))
```

```python

rnn_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python

lstm_model = Sequential()
lstm_model.add(
    Embedding(input_dim=vocal_size, output_dim=embedding_dim, input_length=max_len)
)
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=vocal_size, activation='softmax'))
     
```

```python
lstm_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
epochs = 10
batch_size = 128
```

```python
history_rniin = rnn_model.fit(X_padded,y_one_hot,epochs=epochs,batch_size=batch_size,validation_split=0.1)
```
